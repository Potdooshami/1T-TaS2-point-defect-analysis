# Remove slow-varying back ground

```python
def plot_result(image, background):
    fig, ax = plt.subplots(nrows=1, ncols=3)

    ax[0].imshow(image, cmap='gray')
    ax[0].set_title('Original image')
    ax[0].axis('off')

    ax[1].imshow(background, cmap='gray')
    ax[1].set_title('Background')
    ax[1].axis('off')

    ax[2].imshow(image - background, cmap='gray')
    ax[2].set_title('Result')
    ax[2].axis('off')

    fig.tight_layout()
    return fig,ax
```

```python
from scipy.ndimage import gaussian_filter
import numpy as np
from matplotlib import pyplot as plt
def gaussClean(img,sig):
  sz = list(img.shape)
  sz.append(3)
  imgs = np.zeros(sz)
  imgs[:,:,0] = img
  imgs[:,:,1] = gaussian_filter(img,sigma=sig)
  imgs[:,:,2] = img - gaussian_filter(img,sigma=sig)
  return imgs

def visGaussClean(imgs):
  fig,axs = plt.subplots(1,3,figsize=(15, 5))
  plt.figure(figsize=(20, 20)) # Adjust the size as needed
  tns = ["original","original","raw-gaussed"]
  for ind in range(3):
    axs[ind].imshow(imgs[:,:,ind],cmap='afmhot')#,vmin=-13*p,vmax=18*p)
    axs[ind].set_title(tns[ind])
  return axs

```
